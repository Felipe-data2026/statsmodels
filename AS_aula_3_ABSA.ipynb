# facebook/bart-large-mnni

https://huggingface.co/facebook/bart-large-mnli

O **facebook/bart-large-mnli** é um modelo de linguagem da Meta disponível no Hugging Face, baseado na arquitetura **BART** e treinado para **Natural Language Inference (NLI)**.

Ele é amplamente utilizado para **Zero-Shot Text Classification**, permitindo classificar textos em categorias sem a necessidade de treinamento adicional.

## Principais características

- Classificação *zero-shot* (sem treinamento específico)
- Baseado na arquitetura BART Large
- Treinado com o dataset **MultiNLI**
- Compatível com a biblioteca `transformers`
- Suporta classificação com múltiplas categorias

## Funcionamento

O modelo compara um texto com possíveis categorias ou hipóteses e calcula a probabilidade de cada uma representar melhor o conteúdo analisado.

## Aplicações

- Classificação de documentos
- Detecção de tópicos
- Classificação de intenção (*Intent Classification*)
- Organização automática de textos
- Análise de reviews
- Uso em cenários de **Aspect-Based Sentiment Analysis (ABSA)** quando os aspectos são informados previamente

## Limitações

- Não extrai aspectos automaticamente.
- Não é um modelo especializado em análise de sentimento.
- Pode ter menor precisão que modelos treinados especificamente para uma tarefa.

```python
from transformers import pipeline
```

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

text = "The food was delicious, but the service was terrible."

aspects = ["food", "service"]

for aspect in aspects:
    candidate_labels = [
        f"The sentiment towards {aspect} is positive",
        f"The sentiment towards {aspect} is negative",
        f"The sentiment towards {aspect} is neutral"
    ]

    result = classifier(text, candidate_labels)

    print()
    print(aspect)
    print(result["labels"][0])
    print(result["scores"][0])
    print()
```
